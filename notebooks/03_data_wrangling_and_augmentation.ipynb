# 03 — Data Wrangling and Augmentation

Normalize images and define training-only data augmentation.

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
import numpy as np
```

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
y_train = y_train.ravel()
y_test = y_test.ravel()
```

```python
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0
x_tr, x_val, y_tr, y_val = train_test_split(
    x_train, y_train, test_size=0.1, random_state=42, stratify=y_train
)
print(x_tr.shape, x_val.shape, x_test.shape)
```

```python
augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip('horizontal'),
    tf.keras.layers.RandomTranslation(0.08, 0.08),
    tf.keras.layers.RandomRotation(0.08),
    tf.keras.layers.RandomZoom(0.10),
], name='data_augmentation')
```

```python
sample = tf.expand_dims(x_tr[0], axis=0)
augmented = augmentation(sample, training=True)
print('Original:', sample.shape, 'Augmented:', augmented.shape)
```

Augmentation is applied during training only. Validation and test images remain unmodified apart from normalization.
