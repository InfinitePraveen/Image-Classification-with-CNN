# 02 — EDA and Visualization

Explore class balance, pixel distributions and representative images.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
```

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
class_names = ['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck']
counts = np.bincount(y_train.ravel(), minlength=10)
print(dict(zip(class_names, counts)))
```

```python
sns.barplot(x=class_names, y=counts)
plt.xticks(rotation=45)
plt.title('Training Class Distribution')
plt.ylabel('Images')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.hist(x_train.reshape(-1), bins=50)
plt.title('Pixel Intensity Distribution')
plt.xlabel('Pixel value')
plt.ylabel('Frequency')
plt.show()
```

CIFAR-10 is approximately balanced, making accuracy a useful primary metric alongside the confusion matrix and per-class metrics.
