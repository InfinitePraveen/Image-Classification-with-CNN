# 06 — Inference and Export

Verify the exported model and inspect individual predictions.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
class_names = ['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck']
model_path = Path('../models/cifar10_cnn.keras')
model = tf.keras.models.load_model(model_path)
print(model_path.resolve())
```

```python
(_, _), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
y_test = y_test.ravel()
images = x_test.astype('float32') / 255.0
```

```python
indices = [0, 10, 100, 1000, 5000]
probs = model.predict(images[indices], verbose=0)
for i, p in zip(indices, probs):
    print(f'Actual: {class_names[y_test[i]]:10s} | Predicted: {class_names[p.argmax()]:10s} | Confidence: {p.max():.2%}')
```

```python
metadata = {
    'input_shape': [32, 32, 3],
    'classes': class_names,
    'normalization': 'divide pixels by 255',
    'framework': 'TensorFlow/Keras'
}
import json
with open('../models/model_metadata.json', 'w', encoding='utf-8') as f:
    json.dump(metadata, f, indent=2)
print('Model metadata exported.')
```
