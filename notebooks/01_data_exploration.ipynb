# 01 — Data Exploration

Inspect the CIFAR-10 data, dimensions, class names and sample images.

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
```

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
print(x_train.shape, y_train.shape)
print(x_test.shape, y_test.shape)
```

```python
class_names = ['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck']
print(class_names)
```

```python
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for i, ax in enumerate(axes.flat):
    ax.imshow(x_train[i])
    ax.set_title(class_names[y_train[i, 0].item()])
    ax.axis('off')
plt.tight_layout()
```

### Note
The notebook uses TensorFlow's CIFAR-10 loader for reproducible experimentation. The project dataset is documented as the corresponding Kaggle CIFAR-10 dataset, which contains the same CIFAR-10 classes and standard 50k/10k split.
